## Write an article (≈2 min)

Change the topic to anything you like.

```python
from book_agent import Agent

agent = Agent(autonomous=True, use_images=False)   # images need image search; off for speed
proj = agent.create("How vector databases actually work", mode="article", units=3)
proj.run(progress=print)
```

## Read the finished piece

```python
from IPython.display import Markdown, display

md_path = proj.export("md")
display(Markdown(open(md_path, encoding="utf-8").read()))
```
